# Análise Comparativa dos Gastos com Viagens Oficiais: Governos Bolsonaro e Lula

## Uma Avaliação dos Custos e Padrões de Viagens nos Anos de 2019 a 2024 com Base nos Dados do Portal da Transparência.

**Objetivo do Estudo:**  
Este estudo foi desenvolvido com o intuito de analisar os gastos com viagens oficiais durante os governos de Jair Bolsonaro e Luiz Inácio Lula da Silva. A motivação surgiu a partir de diversas reportagens que indicam um aumento substancial nos custos de viagens durante a gestão atual.

**Metodologia:**  
Para garantir uma comparação justa e tecnicamente fundamentada, foram considerados os anos de 2019 e 2022, representando respectivamente o início e o final do mandato de Bolsonaro, excluindo os anos mais críticos da pandemia. Também foram incluídos os anos de 2023 e 2024, correspondentes aos dois primeiros anos do atual mandato do presidente Lula.

**Objetivo da Análise:**  
A proposta deste estudo não é emitir juízo de valor ou tomar partido político, mas sim contribuir com uma análise baseada em dados públicos. O objetivo é fomentar o debate sobre a alocação de recursos e a transparência nos gastos governamentais.

# Importação de bibliotecas e módulos de apoio

```python
#Importação de bibliotecas principais
import pandas as pd
import os
import warnings
warnings.filterwarnings("ignore")

#Importação de módulos	
from src.processamento import (
    carregar_dados,
    calcular_despesas,
    converter_datas,
    criar_colunas_temporais,
    consolidar_dados,
    preparar_dados_evolucao_top5,
)
from src.plotagem import *
from src.tratamento import filtrar_cargos_relevantes
```

# Preparação e tratamento de dados

```python
# Carregar e processar dados
anos = [2019, 2022, 2023, 2024]
df_viagens = carregar_dados(anos)
df_viagens = calcular_despesas(df_viagens)
df_viagens = converter_datas(df_viagens)
df_viagens = criar_colunas_temporais(df_viagens)
df_consolidado = consolidar_dados(df_viagens)
df_evolucao_top5 = preparar_dados_evolucao_top5(df_viagens)
```

```python
# Filtrar dados relevantes
df_final = filtrar_cargos_relevantes(df_viagens, df_consolidado)
```

**Análise de Viagens por Cargo Público Divididas por Ano:**  
O gráfico apresenta o maior número de viagens realizadas por servidores públicos federais brasileiros, distribuídas por cargo e por ano, com dados extraídos do Portal da Transparência.

**Cargos com Maior Número de Viagens:**

- Informações protegidas por sigilo
- Professor do Magistério Superior
- Professor do Ensino Básico, Técnico e Tecnológico

**Observações:**  
O cargo classificado como "Informações protegidas por sigilo" registrou o maior número de viagens em todos os anos analisados. O pico ocorreu em 2023, com 118.442 viagens, seguido de 2024, com 117.220 viagens. Em 2019, o número foi de 97.558, enquanto em 2022 foi de 100.648. Observa-se um crescimento expressivo ao longo do tempo, exceto em 2019.

Os **Professores do Magistério Superior** apresentaram uma variação significativa. O maior número de viagens foi em 2019 (71.412), com uma queda para 35.321 em 2022, provavelmente devido aos impactos da pandemia. Em 2023, houve uma recuperação para 54.876, mantendo-se próximo de 50.175 em 2024.

Os **Professores do Ensino Básico, Técnico e Tecnológico** tiveram o menor volume de viagens entre os três grupos. O número caiu de 44.157 em 2019 para 33.856 em 2024, com 37.673 em 2023, indicando uma tendência de redução ao longo do tempo.

**Considerações:**  
O gráfico revela padrões distintos de viagens entre os cargos analisados. O dado mais destacado é que a maior parte das viagens realizadas anualmente está associada a cargos classificados como "Informações protegidas por sigilo". Em todos os anos observados, esse grupo superou consideravelmente os demais em número de viagens, com picos como 118.442 viagens em 2023 e 117.220 em 2024.

Esse volume elevado e constante de viagens, cuja natureza não é detalhada devido ao sigilo, levanta questões significativas sobre a transparência na gestão de recursos públicos, especialmente quando comparado com os números de viagens dos professores do magistério superior e do ensino técnico, que apresentam volumes muito inferiores e mais suscetíveis a oscilações.

Os professores, por sua vez, exibiram uma tendência mais volátil, com uma queda acentuada em 2022 e uma leve recuperação nos anos subsequentes, possivelmente refletindo restrições orçamentárias, mudanças de gestão ou políticas de contenção de despesas.

Essa discrepância entre as categorias reforça a necessidade de uma análise crítica sobre os critérios adotados para a autorização de viagens e sobre a própria classificação de sigilo aplicada a uma grande parte desses dados.
